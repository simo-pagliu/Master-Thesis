```python
import numpy as np
import matplotlib.pyplot as plt
```

### Load Following Capability
This is defined as the area of allowable ramp-up/down range over the span of the reactor allowable power

```python
# EPR 1750
# Define ramp rates for each power range segment
P_range = [0.25, 0.6, 1.0] # % P_nom
P_ramp = [2.5, 5.0] # % P_nom/min

# Compute the area under the ramp rate curve
area = (P_range[1] - P_range[0]) * P_ramp[0] + (P_range[2] - P_range[1]) * P_ramp[1]

# Plot
plt.figure(figsize=(5, 5))
plt.step([0] + P_range[:2] + [P_range[2]], [0, P_ramp[0], P_ramp[1], P_ramp[1]], where='post', color='C0', label='Stepwise Ramp Rate')
plt.fill_between([P_range[0], P_range[1]], P_ramp[0], color='C0', alpha=0.3)
plt.fill_between([P_range[1], P_range[2]], P_ramp[1], color='C0', alpha=0.3)
plt.title('EPR 1750 - Power Range and Ramp Rates')
plt.xlabel(r'Power Range ($\%\,P_\mathrm{nom}$)')
plt.ylabel(r'Ramp Rate ($\%\,P_\mathrm{nom}/\mathrm{min}$)')
plt.text(0.5, 5.5, r'Area $= {:.3f}\ (\%\,P_\mathrm{{nom}}^2/\mathrm{{min}})$'.format(area), fontsize=12, ha='center', bbox=dict(facecolor='white', alpha=0.7))
plt.xlim(0, 1)
plt.ylim(0, 6)
plt.grid()
plt.show()
```
