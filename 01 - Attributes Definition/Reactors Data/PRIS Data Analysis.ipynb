```python
df_StartsConstruction["Status"] = "Under Construction"
df_UnderConstruction["Status"] = "Under Construction"

df_PlannedConstruction["Status"] = "Planned Construction"

df_JustConnected["Status"] = "Scheduled Connected"
df_ScheduledConnection["Status"] = "Scheduled Connection"

df_Operational["Status"] = "Operational"

# Unite all DataFrames
df_Complete = pd.concat([df_StartsConstruction, df_UnderConstruction, df_PlannedConstruction, df_JustConnected, df_ScheduledConnection, df_Operational], ignore_index=True)

# Display the final DataFrame
pd.options.display.max_columns = None  # Show all columns
display(df_Complete)
```

```python
# List all Countries
countries = df_Complete["Country"].unique()
display(countries)
```

```python
# Exclude countries known to only work within its orders
To_exclude = ['CHINA', 'RUSSIA', 'IRAN', 'INDIA', 'BELARUS', 'PAKISTAN']
# Exclude this from the df
df_Countries_Filtered = df_Complete[~df_Complete["Country"].isin(To_exclude)]
```

```python
# List all nsss Suppliers
nsss_suppliers = df_Countries_Filtered["nsss Supplier"].unique()
display(nsss_suppliers)
```

```python
operator_country_array = [
    ["JSC ASE", "Russia"],
    ["CNEA", "Argentina"],
    ["AEM", "Russia"],
    ["KWU", "Germany"],
    ["FRAM", "France"],
    ["H/G", "Japan"],
    ["HITACHI", "Japan"],
    ["DHICKOPC", "Korea"],
    ["ŠKODA", "Czech"],
    ["KEPCO", "Korea"],
    ["ORANO", "France"],
    ["WH", "US"],
    ["nan", "Japan (GE Hitachi Nuclear Energy or Toshiba)"],
    ["MHI", "US/Japan"],
    ["SIEMENS", "Germany"],
    ["AECL", "Canada"],
    ["FAEA", "US"],
    ["ACECOWEN", "US/Belgium"],
    ["ACLF", "France/Belgium"],
    ["AEE", "Russia"],
    ["OH/AECL", "Canada"],
    ["ASEASTAL", "Finland"],
    ["WH/MHI", "Japan/US"],
    ["DHICKAEC", "Korea"],
    ["AECL/DHI", "Canada"],
    ["GE", "US"],
    ["S/KWU", "Germany"],
    ["ABB ATOM", "Sweden"],
    ["GETSCO", "US"],
    ["NPC", "UK"],
    ["NNC", "UK"],
    ["PPC", "UK"],
    ["PAIP", "Russia"],
    ["PAA", "Russia"],
    ["B&W", "US"],
    ["CE", "US"]
]
# get those that contain in the second element one of the countries from To_Exclude
filtered_operators = [op for op in operator_country_array if op[1].upper() not in To_exclude]
# filtered_operators = [[op[0].upper(), op[1].upper()] for op in filtered_operators]
print("Operators that are not in the excluded countries:")
for op in filtered_operators:
    print(f" - {op[0]} ({op[1]})")
```

```python
# Filter out operators from excluded countries from the df
df_Countries_Filtered_2 = df_Countries_Filtered[df_Countries_Filtered["nsss Supplier"].isin([op[0] for op in filtered_operators])]
display(df_Countries_Filtered_2)
```

```python
# Only consider those with a grid connection that is either TBD or after 1986 (yyyy-mm format)
def age_limit(val):
    if val == "TBD":
        return True
    try:
        year = int(str(val).split('-')[0])
        return year > 2000
    except Exception:
        return False
# Either apply it to the grid connection or the construction starting date, the result is basically the same
df_Age_Filtered = df_Countries_Filtered_2[df_Countries_Filtered_2["Construction Starts"].apply(age_limit)]
display(df_Age_Filtered)
```

```python
models = df_Age_Filtered["Model"].unique()
print(len(models), "unique models found:")
for m in models:
    print(f"- {m}")
```

```python
models_to_study = ["EPR", "APR-1400", "EPR-1750", "AP-1000", "APWR"]
df_filtered_models = df_Countries_Filtered[df_Countries_Filtered["Model"].isin(models_to_study)]
```

```python
# Let's see each model in more detail
for m in models_to_study:
    display(m)
    display(df_filtered_models[df_filtered_models["Model"] == m])
```
