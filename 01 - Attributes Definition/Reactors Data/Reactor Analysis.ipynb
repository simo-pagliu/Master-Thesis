```python
# European countries
european_countries = [
    "Austria", "Belgium", "Bulgaria", "Croatia", "Cyprus", "Czechia", "Denmark", "Estonia", "Finland", "France",
    "Germany", "Greece", "Hungary", "Iceland", "Ireland", "Italy", "Latvia", "Lithuania", "Luxembourg", "Malta",
    "Netherlands", "Norway", "Poland", "Portugal", "Romania", "Slovakia", "Slovenia", "Spain", "Sweden",
    "Switzerland", "United Kingdom"
]

# Convert to uppercase to match df_Complete country names
european_countries = [c.upper() for c in european_countries]
df_Europe = df_Complete[df_Complete["Country"].isin(european_countries)].reset_index(drop=True)

# Compute average Capacity Factor for operational reactors in Europe
operational_reactors_Europe = df_Europe[df_Europe["Status"] == "Operational"]
capacity_factors = operational_reactors_Europe["Capacity Factor"].replace(["", None, pd.NA], pd.NA).dropna().astype(float)
average_capacity_factor = capacity_factors.mean()
print(f"Average Capacity Factor for operational reactors in Europe: {average_capacity_factor:.2f}%")
# plot histogram of capacity factors
import matplotlib.pyplot as plt
plt.hist(capacity_factors, bins=20,  color=(0.36, 0.54, 0.6), edgecolor='black')
plt.title('Histogram of Capacity Factors for Operational Reactors in Europe')
plt.xlabel('Capacity Factor (%)')
plt.ylabel('Number of Reactors')
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(x)}%'))
plt.xlim(60, 100)
plt.show()  
```

```python
print(capacity_factors.mean())
```

```python
# remove below 50% as outliers
capacity_factors_filtered = capacity_factors[capacity_factors >= 50]
average_capacity_factor_filtered = capacity_factors_filtered.mean()
std_capacity_factor_filtered = capacity_factors_filtered.std()
print(f"Average Capacity Factor for operational reactors in Europe (filtered >=50%): {average_capacity_factor_filtered:.2f}% ± {std_capacity_factor_filtered:.2f}%")
```

```python
# Histogram + Gaussian fit scaled to histogram 
import numpy as np
counts, bin_edges = np.histogram(capacity_factors_filtered, bins=20)
bin_width = bin_edges[1] - bin_edges[0]

plt.figure(figsize=(8, 8))
plt.hist(capacity_factors_filtered, bins=bin_edges, color=(0.36, 0.54, 0.6), edgecolor='black', alpha=0.7)

mu = average_capacity_factor_filtered
sigma = std_capacity_factor_filtered

if sigma > 0 and not np.isnan(mu):
    x = np.linspace(bin_edges[0], bin_edges[-1], 200)
    p = np.exp(-0.5 * ((x - mu) / sigma) ** 2) / (sigma * np.sqrt(2 * np.pi))
    # scale PDF to histogram counts: pdf * N * bin_width
    p_scaled = p * len(capacity_factors_filtered) * bin_width
    plt.plot(x, p_scaled, 'k', linewidth=0.5, linestyle = '--')

# plt.title('Histogram of Capacity Factors for Operational Reactors in Europe (Filtered) with Gaussian Fit')
plt.xlabel('Capacity Factor (%)')
plt.ylabel('Number of reactors')
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{int(x)}%'))
plt.xlim(69, 100)
# highlight mu ± sigma region with shaded band, vertical lines and a horizontal connector
left = mu - sigma
right = mu + sigma

plt.axvspan(left, right, color='gray', alpha=0.2, label=r'$\mu \pm \sigma$ (68%)')
plt.axvline(mu, color='black', linestyle='--', linewidth=1.5)
plt.axvline(left, color='red', linestyle='--', linewidth=1,)
plt.axvline(right, color='red', linestyle='--', linewidth=1,)

# draw a horizontal connector line across the mu±sigma region near the top of the axes
ymin, ymax = plt.ylim()
y_line = ymax * 0.92

# annotate mean and sigma on the plot and show legend
plt.text(0.98, 0.98, f"μ = {mu:.2f}%\nσ = {sigma:.2f}%", transform=plt.gca().transAxes,
         ha='right', va='top', fontsize=12, bbox=dict(facecolor='white', alpha=0.8))

plt.legend()
plt.title("Capacity Factors of Reactors in Europe")
plt.savefig("capacity_factors_reactors_europe.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def extract_year(val):
    if pd.isna(val):
        return np.nan
    s = str(val).strip()
    if s.upper() == "TBD" or s == "":
        return np.nan
    try:
        return int(s.split('-')[0])
    except Exception:
        return np.nan

# parse grid years
grid_years = df_Complete['Grid Connection'].apply(extract_year)

# remove anything with 'TBD' in Grid Connection or Construction Starts
starts_str = df_Complete['Construction Starts'].astype(str).str.strip().str.upper()
mask_no_tbd = (~grid_years.isna()) & (~starts_str.eq('TBD'))

# remove construction/planned statuses
mask_no_construction = ~df_Complete['Status'].astype(str).str.contains(r'Construction|Planned', case=False, na=False)

# final mask
mask = mask_no_tbd & mask_no_construction
df_plot = df_Complete[mask].copy()

if df_plot.empty:
    print("No data to plot after removing 'TBD' and construction/planned entries.")
else:
    years = df_plot['Grid Connection'].apply(extract_year).astype(int)
    counts = years.value_counts().sort_index()
    full_idx = np.arange(counts.index.min(), counts.index.max() + 1)
    counts_full = counts.reindex(full_idx, fill_value=0)
    cumulative = counts_full.cumsum()
    plt.rcParams.update({'font.size': 16})
    plt.figure(figsize=(10, 6))
    plt.plot(full_idx, cumulative, linestyle='-', color='tab:blue')
    plt.fill_between(full_idx, cumulative, color='tab:blue', alpha=0.22, step='pre')

    # Remove scatter points
    # plt.scatter(full_idx, cumulative, color='tab:blue', s=12)  # Commented out

    # Add vertical lines for Chernobyl (1986) and Fukushima (2011)
    plt.axvline(x=1986, color='red', linestyle='--', alpha=0.7, label='Chernobyl (1986)')
    plt.axvline(x=2011, color='orange', linestyle='--', alpha=0.7, label='Fukushima (2011)')

    plt.xlabel('Year of Connection to the Grid')
    plt.ylabel('Number of reactors')
    plt.grid(alpha=0.3)
    plt.ylim(bottom=0)
    plt.xlim(1965, 2024)

    # Set xticks every decade
    decade_ticks = np.arange((full_idx.min() // 10) * 10, full_idx.max() + 1, 10)
    plt.xticks(decade_ticks, rotation=45)

    # Add legend for the vertical lines
    plt.legend()

    plt.tight_layout()
    plt.savefig('reactors_grid_connection_cumulative.png', dpi=300)

    plt.show()

    removed = len(df_Complete) - len(df_plot)
    print(f"Plotted {len(df_plot)} reactors. Removed {removed} rows (TBD or construction/planned).")
```

## Load Following Capabilities (LFC)

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# EPR 1750
# Define ramp rates for each power range segment
P_range = [0.25, 0.6, 1.0] # % P_nom
P_ramp = [2.5, 5.0] # % P_nom/min

# Compute the area under the ramp rate curve
area = (P_range[1] - P_range[0]) * P_ramp[0] + (P_range[2] - P_range[1]) * P_ramp[1]

# Plot
plt.figure(figsize=(5, 5))
plt.step([0] + P_range[:2] + [P_range[2]], [0, P_ramp[0], P_ramp[1], P_ramp[1]], where='post', color='C0', label='Stepwise Ramp Rate')
plt.fill_between([P_range[0], P_range[1]], P_ramp[0], color='C0', alpha=0.3)
plt.fill_between([P_range[1], P_range[2]], P_ramp[1], color='C0', alpha=0.3)
plt.title('EPR 1750 - Power Range and Ramp Rates')
plt.xlabel(r'Power Range ($\chi\,P_\mathrm{nom}$)')
plt.ylabel(r'Ramp Rate ($\%\,P_\mathrm{nom}/\mathrm{min}$)')
plt.text(0.5, 5.5, r'Area $= {:.3f}$'.format(area), fontsize=12, ha='center', bbox=dict(facecolor='white', alpha=0.7))
plt.xlim(0, 1)
plt.ylim(0, 6)
plt.grid()
plt.show()
```

```python
Burnup_reactors = [m(46.5,0), m(55,0), m(50,0), m(50,5), m(55,5), m(52.5,5)]
Efficiency_reactors = [m(0.3515,0), m(0.3617,0), m(0.3235,0), m(0.3448,0), m(0.3461,0), m(0.3080,0)]
SNF_produced = []
for i, Reactor in enumerate(Reactors):
    bu_d_el = Burnup_reactors[i] * Efficiency_reactors[i] # GWd_el/tHM
    bu_d_el_yr = bu_d_el / 365
    SNF_hm = 1 / bu_d_el_yr # tHM/GWe/yr
    SNF_produced.append(SNF_hm * (238+2*16)/238)
    print(f"{Reactor}: {SNF_produced[-1]} t_SNF/GWe/yr")
```

```python

CF_eu_avg = m(81.64,5.52) / 100

for i, tons_SNF_GWe_yr in enumerate(SNF_produced):
    SNF_per_year = tons_SNF_GWe_yr * Power[i] * CF_eu_avg / 1000
    print(f"{SNF_per_year}, relative uncertainty: {SNF_per_year.uncertainty/SNF_per_year.value*100:.2f}%")
```
