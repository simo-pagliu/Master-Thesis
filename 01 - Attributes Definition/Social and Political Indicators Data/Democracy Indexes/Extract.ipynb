```python
# Assuming df is your DataFrame and 'year' is the column with the year information
def weighted_avg(group):
    # For each variable, compute the weighted average across years
    variables = ['v2x_polyarchy', 'v2regdur', 'v2x_jucon', 'v2cltrnslw', 'v2dlengage']
    result = {}
    for var in variables:
        # Drop NaN values and sort by year
        data = group[['year', var]].dropna()
        if len(data) == 0:
            result[var] = np.nan
            continue
        # Sort by year to assign increasing weights
        data = data.sort_values('year')
        values = data[var].values
        weights = np.arange(1, len(values) + 1)
        result[var] = np.average(values, weights=weights)
    return pd.Series(result)

# Apply the function to each country
df_weighted = df.groupby('country_name').apply(weighted_avg).reset_index()
print(df_weighted)
```
